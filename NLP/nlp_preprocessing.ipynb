### Tokenization

```python
import nltk
```

```python

corpus="""Lorem Ipsum is simply dummy text of the printing and typesetting industry.
 Lorem Ipsum has been the industry's standard dummy text ever since the 1500s, 
 when an unknown printer took a galley of type and scrambled it to make a type specimen book. 
 It has survived not only five centuries, but also the leap into electronic typesetting, remaining essentially unchanged. 
 It was popularised in the 1960s with the release of Letraset sheets containing Lorem Ipsum passages, 
 and more recently with desktop publishing software like Aldus PageMaker including versions of Lorem Ipsum.
"""

print(corpus)
```

```python
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize
```

```python
sent_tokens=sent_tokenize(text=corpus,language="english")
print(sent_tokens)
```

```python
from nltk.tokenize import word_tokenize
```

```python
word_tokens=word_tokenize(corpus)
print(word_tokens)
```

### Stemming

#### Porter stemming

```python
from nltk.stem import PorterStemmer
```

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
porter_stem=PorterStemmer()

for word in words:
    print("original word->",word,"   ","stemmed word->",porter_stem.stem(word=word))
```

#### RegexStemmer

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stem=RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stem.stem('eating')
```

#### Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snb_stemmer=SnowballStemmer('english')
```

```python
for word in words:
    print("word-->",word,"  ","porter stemper-->",porter_stem.stem(word),"snowball stemmer-->",snb_stemmer.stem(word))
```

### Lemmitization (gives root word)

#### Wordnet Lemmitization

```python
from nltk.stem import WordNetLemmatizer
```

```python
wrd_lemma=WordNetLemmatizer()
```

```python
nltk.download('wordnet')
```

```python
for word in words:
    print("original word->",word,"   ","lemma word->",wrd_lemma.lemmatize(word,pos='v'))
```

### stopwords

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
words=word_tokenize(corpus)
words
```

```python
stemmed_words=[snb_stemmer.stem(word) for word in words]
stemmed_words
```

```python
compressed_words=[word for word in stemmed_words if word not in set(stopwords.words('english'))]
compressed_words
```

```python
compressed_corpus=" ".join(compressed_words)
compressed_corpus
```
