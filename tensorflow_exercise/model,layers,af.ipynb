### Creating a Sequential Model

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
model=keras.Sequential(
    [
    keras.Input(shape=(784,),name="Input Layer"),
    layers.Dense(64,activation="relu",name="dense layer"),
    layers.Dense(10,activation="softmax",name="output layer")
    ], name="simple mlp"
)

model.summary()
```

### Building Incrementally with add()

```python
model_=keras.Sequential(name="mlp usind incremental add")
model_.add(keras.Input(shape=(784,),name="Input Layer"))
model_.add(layers.Dense(64,activation="relu",name="Dense Layer"))
model_.add(layers.Dense(10,activation="softmax",name="Output Layer"))

model_.summary()
```

### Common Keras Layers

#### Dense Layer

```python
dense_layer=layers.Dense(units=64,activation="relu",name="Dense Layer")
dense_layer
```

```python
output_layer=layers.Dense(units=10,activation="softmax",name="output layer")
output_layer
```

#### Conv2D Layer

```python
conv_layer=layers.Conv2D(filters=32,kernel_size=(3,3),activation="relu",padding="same",name="Convolution Layer")
conv_layer
```

```python
input_size=(28,28,1)

model_conv=keras.Sequential(name="Model with conv layer")


model_conv.add(layers.Input(shape=input_size))
model_conv.add(layers.Conv2D(filters=32,kernel_size=(3,3),activation="relu"))
model_conv.summary()
```

#### MaxPooling2D Layer

```python
model_conv.add(layers.MaxPool2D(pool_size=(2,2)))
model_conv.summary()
```

#### Flatten Layer

```python
model_conv.add(layers.Flatten()) #(batch_size, height, width, channels) would be flattened to (batch_size, height * width * channels).
model_conv.summary()
```

#### Dropout Layer

```python
model_conv_=keras.Sequential([
    layers.Input(shape=input_size),
    layers.Conv2D(filters=32,kernel_size=(2,2),activation="relu"),
    layers.Dropout(rate=0.3),
    layers.MaxPool2D(pool_size=(2,2)),
    layers.Dropout(rate=0.3),
    layers.Flatten(),
    layers.Dense(10,activation="softmax")
])

model_conv_.summary()
```

```python
model__= keras.Sequential([
    layers.Input(shape=(784,)),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.3), # Apply 30% dropout
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.3), # Apply 30% dropout
    layers.Dense(10, activation='softmax')
])

model__.summary()
```

#### Lstm Layer

```python
lstm_layer=layers.LSTM(units=64,name="Lstm Layer")
lstm_layer
```

```python
model_lstm=keras.Sequential(
    [
        layers.Input(shape=(None,10)),
        layers.LSTM(units=32),
        layers.Dense(1,activation="sigmoid")
    ]
)

model_lstm.summary()
```
