```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from sklearn import datasets
import numpy as np
```

```python
X_numpy,y_numpy=datasets.make_classification(n_features=9,n_classes=2,n_informative=2,random_state=1)

X=torch.from_numpy(X_numpy.astype(np.float32))
y=torch.from_numpy(y_numpy.astype(np.float32))

y=y.view(y.shape[0],1)
```

```python
n_samples,n_features=X.shape
input_dim=n_features

output_dim=1
```

```python
class LogisticRegression(nn.Module):
    def __init__(self,input_size,output_size):
        super(LogisticRegression,self).__init__()
        self.lin=nn.Linear(in_features=input_size,out_features=output_size)

    def forward(self,x):
        return self.lin(x)
```

```python
model=LogisticRegression(input_size=input_dim,output_size=output_dim)
criterion=nn.BCELoss()
optimizer=torch.optim.SGD(model.parameters(),lr=0.01)
```

```python
import torch.nn.functional as F
```

```python
for epoch in range(100):

    y_pred=F.sigmoid(model(X))

    l=criterion(y_pred,y)

    l.backward()

    optimizer.step()

    optimizer.zero_grad()

    if epoch%10==0:
        print(f"epoch {epoch+1}: ,loss={l.item():.8f}")
```
