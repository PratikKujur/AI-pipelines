```python
import torch
```

```python
x=torch.randn(3,requires_grad=True)
print(x)
```

forward propagation

```python
y=x+2
print(y)
```

compute loss 

```python
z=y*y*2
z=z.mean()
print(z)
```

backward propagation

```python
z.backward()#dz/dx
print(x.grad)
```

prevent gradient history

```python
x=torch.randn(3,requires_grad=True)
print(x)
```

```python
x.detach()
```

```python
with torch.no_grad():
    y=x+2
    print(y)
```

Training example

```python
weights=torch.ones(4,requires_grad=True)


for epoch in range(3):
    model_output=(weights*3).sum()

    model_output.backward()

    print(weights.grad)

    weights.grad.zero_()
```
