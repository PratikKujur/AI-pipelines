```python
import torch
import torchvision
from torch.utils.data import Dataset, DataLoader
import numpy as np
import math
from sklearn.datasets import load_wine
```

```python
class WineDateset(Dataset):
    def __init__(self,transform=None):
        wine = load_wine()
        self.X,self.y=(wine.data),(wine.target)
        self.n_samples=self.X.shape[0]
        self.transform=transform


    def __getitem__(self,index):
        sample= self.X[index],self.y[index]

        if self.transform:
            sample=self.transform(sample)
        return sample
    
    def __len__(self):
        return self.n_samples
```

```python
class ToTensor:
    def __call__(self,sample):
        inputs, targets = sample
        inputs = torch.from_numpy(inputs)
        targets = torch.tensor(targets) if not isinstance(targets, np.ndarray) else torch.from_numpy(targets)
        return inputs, targets
```

```python
dataset=WineDateset(transform=None)
first_data=dataset[0]
features,labels=first_data

print(type(features), type(labels))
```

```python
class MulTransform:
    def __init__(self,factor):
        self.factor=factor
    
    def __call__(self,sample):
        inputs,target=sample
        inputs*=self.factor
        return inputs,target
```

```python
dataset=WineDateset(transform=ToTensor())
first_data=dataset[0]
features,labels=first_data
print("features before transformation-->",features)
print(type(features), type(labels))

composed=torchvision.transforms.Compose([ToTensor(),MulTransform(4)])
dataset=WineDateset(transform=composed)
first_data=dataset[0]
features,labels=first_data
print("features after transformation-->",features)
print(type(features), type(labels))
```
